## Part A: Scikit-learn preprocessing

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

categorical_cols = [
    "quarter",
    "department",
    "day"
]

numeric_cols = [
    "team",
    "targeted_productivity",
    "smv",
    "wip",
    "over_time",
    "incentive",
    "idle_time",
    "idle_men",
    "no_of_style_change",
    "no_of_workers",
    "year",
    "month",
    "day_of_month"
]
```

```python
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])
```

```python
preprocessor = ColumnTransformer([
    ("num", numeric_pipeline, numeric_cols),
    ("cat", categorical_pipeline, categorical_cols)
])
```

### Removing NAN values 

```python
# Start again from the original training and testing data

X_reg_train_manual = X_reg_train.copy()
X_reg_test_manual = X_reg_test.copy()

# 1. Fill missing numerical values using training medians
numeric_medians = X_reg_train_manual[numeric_cols].median()

X_reg_train_manual[numeric_cols] = (
    X_reg_train_manual[numeric_cols]
    .fillna(numeric_medians)
)

X_reg_test_manual[numeric_cols] = (
    X_reg_test_manual[numeric_cols]
    .fillna(numeric_medians)
)

# 2. Remove extra spaces from categorical values
for col in categorical_cols:
    X_reg_train_manual[col] = X_reg_train_manual[col].str.strip()
    X_reg_test_manual[col] = X_reg_test_manual[col].str.strip()

# 3. One-hot encode categorical columns
X_reg_train_manual = pd.get_dummies(
    X_reg_train_manual,
    columns=categorical_cols,
    dtype=float
)

X_reg_test_manual = pd.get_dummies(
    X_reg_test_manual,
    columns=categorical_cols,
    dtype=float
)

# 4. Make sure train and test have exactly the same columns
X_reg_test_manual = X_reg_test_manual.reindex(
    columns=X_reg_train_manual.columns,
    fill_value=0
)

# 5. Calculate mean and standard deviation ONLY from training data
train_mean = X_reg_train_manual.mean()
train_std = X_reg_train_manual.std()

# Avoid division by zero for constant columns
train_std[train_std == 0] = 1

# 6. Scale using training statistics
X_reg_train_manual = (
    X_reg_train_manual - train_mean
) / train_std

X_reg_test_manual = (
    X_reg_test_manual - train_mean
) / train_std
```

### FOR LOGISTIC REGRESSION 

### Make copies

```python
X_cls_train_manual = X_cls_train.copy()
X_cls_test_manual = X_cls_test.copy()
```

```python
numeric_medians_cls = X_cls_train_manual[numeric_cols].median()

X_cls_train_manual[numeric_cols] = (
    X_cls_train_manual[numeric_cols]
    .fillna(numeric_medians_cls)
)

X_cls_test_manual[numeric_cols] = (
    X_cls_test_manual[numeric_cols]
    .fillna(numeric_medians_cls)
)
```

```python
print("Missing values in train:",
    X_cls_train_manual.isna().sum().sum())

print("Missing values in test:",
    X_cls_test_manual.isna().sum().sum())
```

```python
for col in categorical_cols:
    X_cls_train_manual[col] = X_cls_train_manual[col].str.strip()
    X_cls_test_manual[col] = X_cls_test_manual[col].str.strip()
```

```python
X_cls_train_manual = pd.get_dummies(
    X_cls_train_manual,
    columns=categorical_cols,
    dtype=float
)

X_cls_test_manual = pd.get_dummies(
    X_cls_test_manual,
    columns=categorical_cols,
    dtype=float
)
```

```python
X_cls_test_manual = X_cls_test_manual.reindex(
    columns=X_cls_train_manual.columns,
    fill_value=0
)

print("Train shape:", X_cls_train_manual.shape)
print("Test shape :", X_cls_test_manual.shape)

print(
    "Same columns:",
    list(X_cls_train_manual.columns) ==
    list(X_cls_test_manual.columns)
)
```

### Calculate training mean and standard deviation

```python
train_mean_cls = X_cls_train_manual.mean()
train_std_cls = X_cls_train_manual.std()

# Avoid division by zero for constant columns
train_std_cls[train_std_cls == 0] = 1
```

```python
X_cls_train_manual = (
    X_cls_train_manual - train_mean_cls
) / train_std_cls

X_cls_test_manual = (
    X_cls_test_manual - train_mean_cls
) / train_std_cls
```

### Checking for NAN

```python
print("NaN in train:",
        X_cls_train_manual.isna().sum().sum())

print("NaN in test:",
        X_cls_test_manual.isna().sum().sum())

print("Train shape:", X_cls_train_manual.shape)
print("Test shape :", X_cls_test_manual.shape)
```

### Converting into numpy 

```python
X_cls_train_np = X_cls_train_manual.to_numpy()
X_cls_test_np = X_cls_test_manual.to_numpy()

y_cls_train_np = y_cls_train.to_numpy()
y_cls_test_np = y_cls_test.to_numpy()
```

```python
print("X train:", X_cls_train_np.shape)
print("X test :", X_cls_test_np.shape)
print("y train:", y_cls_train_np.shape)
print("y test :", y_cls_test_np.shape)
```

```python
X_cls_train_np = np.c_[
    np.ones(X_cls_train_np.shape[0]),
    X_cls_train_np
]

X_cls_test_np = np.c_[
    np.ones(X_cls_test_np.shape[0]),
    X_cls_test_np
]
```

```python
print("X train:", X_cls_train_np.shape)
print("X test :", X_cls_test_np.shape)
```

### Creating the Sigmoid Function 

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

theta_cls = np.zeros(X_cls_train_np.shape[1])

learning_rate = 0.01
iterations = 5000
```

### Gradient descent

```python
start_time = time.perf_counter()

n = X_cls_train_np.shape[0]

for i in range(iterations):

    # Linear score
    z = X_cls_train_np @ theta_cls

    # Convert score to probability
    prob = sigmoid(z)

    # Calculate gradient
    gradient = (
        X_cls_train_np.T @ (prob - y_cls_train_np)
    ) / n

    # Update coefficients
    theta_cls = theta_cls - learning_rate * gradient

cls_manual_train_time = time.perf_counter() - start_time

print("Training time:", cls_manual_train_time)
print("Theta shape:", theta_cls.shape)
```

### Predict probabilities

```python
start_time = time.perf_counter()

z_test = X_cls_test_np @ theta_cls

y_cls_prob_manual = sigmoid(z_test)

cls_manual_prediction_time = time.perf_counter() - start_time

print("Prediction time:", cls_manual_prediction_time)
print("First 10 probabilities:")
print(y_cls_prob_manual[:10])
```

### Convert Probabilities to Predictions

```python
y_cls_pred_manual = (
    y_cls_prob_manual >= 0.5
).astype(int)

print("First 10 predictions:")
print(y_cls_pred_manual[:10])
```

### Calculating Manual Accuracy 

```python
correct = np.sum(
    y_cls_test_np == y_cls_pred_manual
)

total = len(y_cls_test_np)

cls_manual_accuracy = correct / total

print("Manual Accuracy:", cls_manual_accuracy)
```

### Calculate the confusion counts

```python
tp = np.sum(
    (y_cls_test_np == 1) &
    (y_cls_pred_manual == 1)
)

tn = np.sum(
    (y_cls_test_np == 0) &
    (y_cls_pred_manual == 0)
)

fp = np.sum(
    (y_cls_test_np == 0) &
    (y_cls_pred_manual == 1)
)

fn = np.sum(
    (y_cls_test_np == 1) &
    (y_cls_pred_manual == 0)
)

print("TP:", tp)
print("TN:", tn)
print("FP:", fp)
print("FN:", fn)
```

```python
cls_manual_precision = tp / (tp + fp)
print("Manual Precision:", cls_manual_precision)
```

```python
cls_manual_recall = tp / (tp + fn)
print("Manual Recall:", cls_manual_recall)
```

```python
cls_manual_f1 = (
    2 * cls_manual_precision * cls_manual_recall
) / (
    cls_manual_precision + cls_manual_recall
)

print("Manual F1:", cls_manual_f1)
```

### Saving Final Classification Results

```python
cls_manual_results = pd.DataFrame({
    "Model": ["Manual Logistic Regression"],
    "Accuracy": [cls_manual_accuracy],
    "Precision": [cls_manual_precision],
    "Recall": [cls_manual_recall],
    "F1": [cls_manual_f1],
    "Training Time": [cls_manual_train_time],
    "Prediction Time": [cls_manual_prediction_time]
})

cls_manual_results
```

## Part C - Comparison and Optimization

### Optimizing with learning rate 0.01

```python
theta_cls_improved = np.zeros(X_cls_train_np.shape[1])

learning_rate_improved = 0.01
iterations_improved = 10000

start_time = time.perf_counter()

n = X_cls_train_np.shape[0]

for i in range(iterations_improved):

    z = X_cls_train_np @ theta_cls_improved

    prob = sigmoid(z)

    gradient = (
        X_cls_train_np.T @ (prob - y_cls_train_np)
    ) / n

    theta_cls_improved = (
        theta_cls_improved -
        learning_rate_improved * gradient
    )

cls_improved_train_time = time.perf_counter() - start_time

print("Improved training time:", cls_improved_train_time)
```

```python
start_time = time.perf_counter()

z_test_improved = X_cls_test_np @ theta_cls_improved

y_cls_prob_improved = sigmoid(z_test_improved)

y_cls_pred_improved = (
    y_cls_prob_improved >= 0.5
).astype(int)

cls_improved_prediction_time = time.perf_counter() - start_time

print("Improved prediction time:",
        cls_improved_prediction_time)
```

```python
correct_improved = np.sum(
    y_cls_test_np == y_cls_pred_improved
)

cls_improved_accuracy = (
    correct_improved / len(y_cls_test_np)
)

print("Improved Accuracy:", cls_improved_accuracy)
```

### Optimizing with learning rate 0.05 

```python
theta_cls_improved = np.zeros(X_cls_train_np.shape[1])

learning_rate_improved = 0.05
iterations_improved = 10000

start_time = time.perf_counter()

n = X_cls_train_np.shape[0]

for i in range(iterations_improved):

    z = X_cls_train_np @ theta_cls_improved

    prob = sigmoid(z)

    gradient = (
        X_cls_train_np.T @ (prob - y_cls_train_np)
    ) / n

    theta_cls_improved = (
        theta_cls_improved -
        learning_rate_improved * gradient
    )

cls_improved_train_time = time.perf_counter() - start_time

print("Improved training time:", cls_improved_train_time)
```

```python
start_time = time.perf_counter()

z_test_improved = X_cls_test_np @ theta_cls_improved

y_cls_prob_improved = sigmoid(z_test_improved)

y_cls_pred_improved = (
    y_cls_prob_improved >= 0.5
).astype(int)

cls_improved_prediction_time = time.perf_counter() - start_time

print("Improved prediction time:",
        cls_improved_prediction_time)
```

```python
correct_improved = np.sum(
    y_cls_test_np == y_cls_pred_improved
)

cls_improved_accuracy = (
    correct_improved / len(y_cls_test_np)
)

print("Improved Accuracy:", cls_improved_accuracy)
```

The experiment demonstrates that machine learning algorithms such as Linear Regression and Logistic Regression can be implemented from scratch using NumPy and Pandas while achieving performance comparable to Scikit-learn.
