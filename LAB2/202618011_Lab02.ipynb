### Task 2 - Vectorized Arithmetic and Linear Algebra 

```python
import numpy as np

# 1. Create two matrices
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)


# 2. Matrix Addition
addition = A + B

print("\nMatrix Addition (A + B):")
print(addition)

# 3. Element-wise Multiplication
element_multiplication = A * B

print("\nElement-wise Multiplication (A * B):")
print(element_multiplication)

# 4. Matrix Multiplication

matrix_multiplication = A @ B

print("\nMatrix Multiplication (A @ B):")
print(matrix_multiplication)

# 5. Transpose of a square matrix
transpose_A = A.T

print("\nTranspose of A:")
print(transpose_A)


# 6. Determinant of the square matrix 

determinant_A = np.linalg.det(A)

print("\nDeterminant of A:")
print(determinant_A)


# 7. Find inverse if the matrix is invertible

if not np.isclose(determinant_A, 0):

    inverse_A = np.linalg.inv(A)

    print("\nInverse of A:")
    print(inverse_A)

    
    # 8. Verify A @ A^-1 = Identity Matrix
    verification = A @ inverse_A

    print("\nA @ A^-1:")
    print(verification)

    print("\nIs A @ A^-1 equal to the identity matrix?")
    print(np.allclose(verification, np.eye(A.shape[0])))

else:
    print("\nMatrix A is not invertible because its determinant is 0.")
```

### Task 3 - Normal Distribution and Histogram

```python
import numpy as np
import matplotlib.pyplot as plt

# Set random seed for reproducibility
np.random.seed(42)

# 1. Generate 1000 values from a normal distribution
chosen_mean = 50
chosen_std = 10

data = np.random.normal(
    loc=chosen_mean,
    scale=chosen_std,
    size=1000
)

print("Chosen Mean:", chosen_mean)
print("Chosen Standard Deviation:", chosen_std)

# 2. Find sample mean and sample standard deviation
sample_mean = np.mean(data)

# ddof=1 calculates SAMPLE standard deviation
sample_std = np.std(data, ddof=1)
print("\nSample Mean:", sample_mean)
print("Sample Standard Deviation:", sample_std)


# 3. Compare sample values with chosen values
print("\nComparison:")
print("Difference in Mean:", sample_mean - chosen_mean)
print("Difference in Standard Deviation:", sample_std - chosen_std)

print("\nThe sample mean is close to the chosen mean.")
print("The sample standard deviation is close to the chosen standard deviation.")


# 4. Plot Histogram
plt.figure(figsize=(8, 5))
plt.hist(data, bins=40, edgecolor="black")
plt.title("Histogram of Normally Distributed Data")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.savefig("D:\\DAUSem1\\ML\\Figures\\Histogram.png")
plt.show()
```

### Task 5 - Filtering and Querying

```python
import pandas as pd

# Load the Titanic dataset
df = pd.read_csv("D:\\DAUSem1\\ML\\Lab Assingments\\Dataset\\titanic.csv")

# 1. Male passengers older than 50
male_older_50 = df[(df["Sex"] == "male") & (df["Age"] > 50)]
print("1. Male passengers older than 50:", len(male_older_50))


# 2. Female first-class passengers
female_first_class = df[(df["Sex"] == "female") & (df["Pclass"] == 1)]

total_female_first_class = len(female_first_class)
survival_percentage = female_first_class["Survived"].mean() * 100

print("\n2. Female first-class passengers:", total_female_first_class)
print("   Percentage survived:", survival_percentage, "%")


# 3. Age 20-40, Fare above overall median, and survived
overall_median_fare = df["Fare"].median()

age_20_40_high_fare_survived = df[
    (df["Age"].between(20, 40)) &
    (df["Fare"] > overall_median_fare) &
    (df["Survived"] == 1)
]

print("\n3. Age 20-40, Fare above overall median, and survived:",
      len(age_20_40_high_fare_survived))


# 4. Travelling alone, age < 30, and did not survive
travelling_alone = df[
    (df["SibSp"] == 0) &
    (df["Parch"] == 0) &
    (df["Age"] < 30) &
    (df["Survived"] == 0)
]

print("\n4. Travelling alone, age < 30, and did not survive:",
      len(travelling_alone))


# 5. Embarked = S, Pclass 2 or 3,
#    Fare above Southampton median
southampton = df[df["Embarked"] == "S"]
southampton_median_fare = southampton["Fare"].median()

southampton_filter = df[
    (df["Embarked"] == "S") &
    (df["Pclass"].isin([2, 3])) &
    (df["Fare"] > southampton_median_fare)
]

print("\n5. Embarked='S', Pclass 2 or 3, and Fare above Southampton median:",
      len(southampton_filter))
```

### Task 6 - groupby() and Aggregation 

```python
import pandas as pd

# Load the Titanic dataset
df = pd.read_csv("D:\\DAUSem1\\ML\\Lab Assingments\\Dataset\\titanic.csv")


# 1. Survival rate by Sex
print("SURVIVAL RATE BY SEX :")
survival_by_sex = df.groupby("Sex")["Survived"].mean() * 100
print(survival_by_sex)


# 2. Survival rate by Pclass
print("\nSURVIVAL RATE BY PCLASS :")
survival_by_pclass = df.groupby("Pclass")["Survived"].mean() * 100
print(survival_by_pclass)


# 3. Average Age and Fare by Pclass
print("\nAVERAGE AGE AND FARE BY PCLASS :")
avg_age_fare = df.groupby("Pclass")[["Age", "Fare"]].mean()
print(avg_age_fare)


# 4. Passenger count and survival rate by Sex-Pclass
print("\nPASSENGER COUNT AND SURVIVAL RATE BY SEX-PCLASS :")
sex_pclass = df.groupby(["Sex", "Pclass"]).agg(
    Passenger_Count=("PassengerId", "count"),
    Survival_Rate=("Survived", "mean")
)

sex_pclass["Survival_Rate"] = sex_pclass["Survival_Rate"] * 100

print(sex_pclass)


# 5. Passenger count, average Fare, and survival rate by Embarked
print("\nPASSENGER COUNT, AVERAGE FARE, AND SURVIVAL RATE BY EMBARKED :")
embarked_stats = df.groupby("Embarked").agg(
    Passenger_Count=("PassengerId", "count"),
    Average_Fare=("Fare", "mean"),
    Survival_Rate=("Survived", "mean")
)

embarked_stats["Survival_Rate"] = embarked_stats["Survival_Rate"] * 100

print(embarked_stats)
```

### Task 7 - Missing Values and Fare Outliers 


```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Load Titanic dataset
df = pd.read_csv("D:\\DAUSem1\\ML\\Lab Assingments\\Dataset\\titanic.csv")


# 1. MISSING-VALUE COUNT AND PERCENTAGE FOR EVERY COLUMN

missing_count = df.isnull().sum()
missing_percentage = (missing_count / len(df)) * 100

missing_table = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage": missing_percentage
})

print("MISSING VALUES : ")
print(missing_table)


# Plot missing-value counts
plt.figure(figsize=(10, 5))
missing_count.plot(kind="bar")
plt.title("Missing Values by Column")
plt.xlabel("Columns")
plt.ylabel("Missing Count")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("D:\\DAUSem1\\ML\\Figures\\missing_value.png")
plt.show()

# 2. FILL MISSING AGE VALUES WITH MEAN AGE

print("\nAGE MISSING VALUES BEFORE AND AFTER MEAN IMPUTATION :")

print("Before:", df["Age"].isnull().sum())

mean_age = df["Age"].mean()
df["Age"] = df["Age"].fillna(mean_age)

print("After:", df["Age"].isnull().sum())
# 3. IMPUTATION USING MEAN, MEDIAN, MODE AND RANDOM VALUE

# Create separate copies so we can compare different methods
# Mean imputation
df_mean = pd.read_csv("D:\\DAUSem1\\ML\\Lab Assingments\\Dataset\\titanic.csv")
df_mean["Age"] = df_mean["Age"].fillna(df_mean["Age"].mean())

# Median imputation
df_median = pd.read_csv("D:\\DAUSem1\\ML\\Lab Assingments\\Dataset\\titanic.csv")
df_median["Age"] = df_median["Age"].fillna(df_median["Age"].median())

# Mode imputation
df_mode = pd.read_csv("D:\\DAUSem1\\ML\\Lab Assingments\\Dataset\\titanic.csv")
df_mode["Age"] = df_mode["Age"].fillna(df_mode["Age"].mode()[0])

# Random-value imputation
df_random = pd.read_csv("D:\\DAUSem1\\ML\\Lab Assingments\\Dataset\\titanic.csv")

missing_age = df_random["Age"].isnull()

# Randomly select existing Age values
random_values = np.random.choice(
    df_random["Age"].dropna(),
    size=missing_age.sum()
)

df_random.loc[missing_age, "Age"] = random_values


print("\nIMPUTATION RESULTS :")

print("Original missing Age values:",
      pd.read_csv("D:\\DAUSem1\\ML\\Lab Assingments\\Dataset\\titanic.csv")["Age"].isnull().sum())

print("After Mean imputation:",
      df_mean["Age"].isnull().sum())

print("After Median imputation:",
      df_median["Age"].isnull().sum())

print("After Mode imputation:",
      df_mode["Age"].isnull().sum())

print("After Random-value imputation:",
      df_random["Age"].isnull().sum())


# 4. FARE OUTLIERS USING IQR
Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Find outliers
fare_outliers = df[
    (df["Fare"] < lower_bound) |
    (df["Fare"] > upper_bound)
]

print("\nFARE OUTLIERS : ")
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Fare Outliers:", len(fare_outliers))
```

### Task 8 - Features and Pivot Table

```python
import pandas as pd

# Load Titanic dataset
df = pd.read_csv("D:\\DAUSem1\\ML\\Lab Assingments\\Dataset\\titanic.csv")

# 1. CREATE FamilySize
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1


# 2. CREATE IsAlone

df["IsAlone"] = (df["FamilySize"] == 1).astype(int)


print("NEW FEATURES :")
print(df[["SibSp", "Parch", "FamilySize", "IsAlone"]].head())

# 3. CREATE PIVOT TABLE

pivot_table = pd.pivot_table(
    df,
    values="Survived",
    index="Sex",
    columns="Pclass",
    aggfunc="mean"
)

print("\nPIVOT TABLE: SURVIVAL RATE :")
print(pivot_table)

# 4. FIND HIGHEST AND LOWEST GROUP

highest_group = pivot_table.stack().idxmax()
highest_rate = pivot_table.stack().max()

lowest_group = pivot_table.stack().idxmin()
lowest_rate = pivot_table.stack().min()


print("\nHIGHEST SURVIVAL GROUP :")
print("Sex:", highest_group[0])
print("Pclass:", highest_group[1])
print("Survival Rate:", highest_rate * 100, "%")


print("\nLOWEST SURVIVAL GROUP :")
print("Sex:", lowest_group[0])
print("Pclass:", lowest_group[1])
print("Survival Rate:", lowest_rate * 100, "%")
```

## Task 9 - Visualizations and Observations

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Load Titanic dataset
df = pd.read_csv("D:\\DAUSem1\\ML\\Lab Assingments\\Dataset\\titanic.csv")

# 1. CORRELATION HEATMAP
numeric_cols = [
    "Survived",
    "Pclass",
    "Age",
    "SibSp",
    "Parch",
    "Fare"
]

# Convert selected columns to numeric
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

correlation = df[numeric_cols].corr()

print("CORRELATION MATRIX :")
print(correlation)


# Plot correlation heatmap
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.tight_layout()
plt.savefig("D:\\DAUSem1\\ML\\Figures\\correlation.png")
plt.show()


# Find strongest positive and negative relationships
corr_pairs = correlation.where(
    ~np.tril(np.ones(correlation.shape)).astype(bool)
)

corr_pairs = corr_pairs.stack().sort_values()

print("\nSTRONGEST NEGATIVE RELATIONSHIP :")
negative_pair = corr_pairs.iloc[0]
print(negative_pair)

print("\n STRONGEST POSITIVE RELATIONSHIP ")
positive_pair = corr_pairs.iloc[-1]
print(positive_pair)

# 2. SURVIVAL RATE BY SEX
survival_by_sex = df.groupby("Sex")["Survived"].mean() * 100

print("\nSURVIVAL RATE BY SEX :")
print(survival_by_sex)


# Plot survival rate by Sex
plt.figure(figsize=(6, 5))

survival_by_sex.plot(kind="bar")

plt.title("Survival Rate by Sex")
plt.xlabel("Sex")
plt.ylabel("Survival Rate (%)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("D:\\DAUSem1\\ML\\Figures\\Survival_rate.png")
plt.show()

# 3. AGE VS FARE BY SURVIVAL
plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=df,
    x="Age",
    y="Fare",
    hue="Survived",
    alpha=0.7
)

plt.title("Age vs Fare by Survival")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.legend(title="Survived")
plt.tight_layout()
plt.savefig("D:\\DAUSem1\\ML\\Figures\\age_vs_fare.png")
plt.show()

# 4. NUMERICAL RESULTS
male_survival = df[df["Sex"] == "male"]["Survived"].mean() * 100
female_survival = df[df["Sex"] == "female"]["Survived"].mean() * 100

average_age = df["Age"].mean()
average_fare = df["Fare"].mean()

print("\n----- NUMERICAL RESULTS -----")

print("Male survival rate:", round(male_survival, 2), "%")
print("Female survival rate:", round(female_survival, 2), "%")
print("Average Age:", round(average_age, 2))
print("Average Fare:", round(average_fare, 2))

```
